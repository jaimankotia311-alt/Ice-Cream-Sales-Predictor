```python
import pandas as pd
```

```python
data = pd.read_csv("Ice_cream selling data.csv")
```

```python
data.head()
```

```python
# plr = poly linear regression

# slr = low accuracy --> improve --> plr
```

```python
'''
plr
minimum 2 columns
x -> independent
y -> dependent
data behaviour should be curved line, should not be linear

'''
```

```python
x = data.iloc[:,:-1]
x.head()
```

```python
y = data["Ice Cream Sales (units)"]
y.head()
```

```python
'''
import matplotlib.pyplot as plt 
matplot = the main library
pyplot = module inside matplot that is used for ploting
'''
```

```python
import matplotlib.pyplot as plt 
```

```python
plt.scatter(data["Temperature (°C)"],data["Ice Cream Sales (units)"])
plt.title("Tempperature VS Revenue")
plt.xlabel("Temperature")
plt.ylabel("Revenue")
```

```python
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score 
```

```python
x_train , x_test, y_train , y_test = train_test_split(x,y, test_size=0.2, random_state=42)
```

```python
x_train.shape
```

```python
y_train.shape
```

```python
model = LinearRegression()
model.fit(x_train,y_train)
```

```python
y_pred = model.predict(x_test)
```

```python
r2_score(y_test,y_pred)*100    # testing accuracy
```

```python
r2_score(y_train , model.predict(x_train))*100  # traiinig aaccuracy
```

```python
plt.scatter (x_train,y_train, color = "red", label="training data")
plt.scatter (x_test,y_test, color = "blue", label="testing data")
plt.plot(x_test, y_pred, label="BFT",color= "green")
plt.title("BFT")
plt.legend()
plt.show()
```

```python
plt.scatter (x_train,y_train, color = "red", label="training data")
plt.scatter (x_test,y_test, color = "blue", label="testing data")
plt.plot(x_train , model.predict(x_train), label="BFT",color= "green")
plt.legend()
plt.show()
```
